```python
# replacing the missing value with mean value
air_quality_data = air_quality_data.fillna(air_quality_data.mean(numeric_only=True))
```

**Forecasting with FB Prophet Algorithms**

FB Prophet Documentation : https://facebook.github.io/prophet/docs/quick_start.html

```python
# converting the data from DD/MM/YYYY to YYYY/MM/DD

data_info = pd.to_datetime(air_quality_data['Date'])
print(data_info)
```

```python
time_info = air_quality_data['Time']
```

```python
print(time_info)
```

```python
time_info= time_info.apply(lambda x: x.replace('.',','':'))
```

```python
print(time_info)
```

```python
print(type(data_info))
print(type(time_info))
```

```python
#combining  2 series to a pandas dataframe
data_time = pd.concat([data_info, time_info] , axis =1)
```

```python
data_time.head()
```

```python
data_time.shape
```

```python
# combing date and time (mention in FB Prophet documentation )

data_time['ds'] = data_time['Date']. astype(str)+' '+data_time['Time'].astype(str)
```

```python
data_time.head()
```

```python
data_time.info()
```

we have to convert ds from object datatype to date time format

```python
data = pd.DataFrame()
```

```python
data['ds'] = pd.to_datetime(data_time['ds'])
```

```python
data.head()
```

```python
data['y'] =air_quality_data['RH']  # RH = Relative Humidity(dataset info)
```

```python
data.head()
```

```python
pip install prophet
```

```python
from prophet import Prophet
```

# eda

```python
#training the prophet model

model = Prophet()
model.fit(data)
```

```python
future = model.make_future_dataframe(periods= 365, freq= 'H')
future.tail()
```

this is for predicting future
and 'H' --> forecast for hourly data itself  

```python
forecast = model.predict(future)
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail() # yhat --> is a predicting value
```

```python
fig1 = model.plot(forecast) # graph represent trends of data
# line represent what the trend is going to be in future
```

```python
fig2 = model.plot_components(forecast)
```

 Above plot Graph is only for RH you can also plot it for various components in air for prediction


**Train -Test Split (Prophet model)**


```python
# FB is a lib in python(ML) used in performing the task like Regression, Classification , dimension reduction
# , train - test split
from prophet import Prophet
```
